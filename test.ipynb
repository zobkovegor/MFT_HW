```python
from src.data_loader import prepare_futures_data
from src.bar_builder import build_15min_bars
from src.signal import add_high_volume_breakout_signal
import pandas as pd

df = prepare_futures_data("data/RB9999.XSGE.csv")
df_15m = build_15min_bars(df)
df_1 = add_high_volume_breakout_signal(df_15m)
start_date_train = pd.Timestamp("2015-01-01")
end_date_train = pd.Timestamp("2018-12-31 23:59:59")

df_1["next_return_15m"] = (
    df_1
    .groupby("session_block")["close"]
    .transform(lambda x: x.shift(-1) / x - 1)
)

rb_train_trades = df_1.loc[
    (df_1["start"] >= start_date_train)
    & (df_1["start"] <= end_date_train)
    & (df_1["strategy_signal"] != 0)
    & df_1["next_return_15m"].notna()
].copy()

print("Количество сделок:", len(rb_train_trades))
```
